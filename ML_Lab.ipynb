# Task 4: The 5 Vs of This Dataset

* **Volume**: The file size on disk is almost 15 KB. At this scale, it is not "big data" by itself; it fits entirely into local memory and processes in milliseconds.
* **Velocity**: Receiving 50 applications a day yields about $50 \times 365 = 18,250$ rows per year which would likely outgrow a standard spreadsheet tool and transition to a relational database or analytical pipeline once reached hundreds of thousands of rows or require concurrent multiuser write operations.
* **Variety**: PDF or images of identification documents, bank statement PDFs
* **Veracity**: The analsis shows missing data fields in the major columons of income and collateral columns as well as duplicate entries that distort raw distributions.
* **Value**: The bank could analyze the relationship between collateral provision and default rates to optimize lending policies for instance, adjusting risk premiums or tightening approval thresholds for applicants who lack collateral.

# Task 5 Conditional Probability, Revisited

```python
p_default = (df["defaulted"] == "Yes").mean()
print("P(Default):", p_default)
rate_by_collateral = df.groupby("has_collateral")["defaulted"].apply(
lambda x: (x == "Yes").mean()
)
print(rate_by_collateral)
```
